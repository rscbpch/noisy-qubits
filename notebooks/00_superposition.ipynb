# 00 · Superposition / quantum randomness
Template notebook: import from the `noisy_qubits` package, then explain and plot.

```python
from noisy_qubits.algorithms import superposition

r = superposition.run(n_qubits=4, shots=1024)
print("uniformity score:", round(r.score, 3))
r.circuit.draw(output="mpl")
```

```python
from qiskit.visualization import plot_histogram
plot_histogram(r.counts, title="Superposition measurement counts")
```
